  <h2 style="color:#2c3e50; font-family:'Helvetica Neue',sans-serif; font-size:28px; margin-bottom:20px;">
    Introduction:
  </h2>
<p style="text-indent:30px; color: #34495e; font-family: 'Arial', sans-serif; font-size: 15px; margin-top: 18px; line-height:1.6;">
    Protein–protein interaction (PPI) networks and domain–domain interaction data can be used to infer likely epitope regions by identifying which structural surfaces of a receptor are repeatedly contacted, monitored, or scaffolded by specific classes of binding partners (Mahler et al., 2003). Interactors involved in ER quality control tend to engage extracellular or luminal domains, revealing structurally sensitive, conformationally dynamic regions that often serve as antibody epitopes (Araki & Nagata, 2011). In contrast, trafficking and cytoskeletal partners bind intracellular motifs, helping exclude those regions as potential epitope sites (Mahler et al., 2003). When combined with domain-level mapping of both NR1 and its interactors, these data suggest which receptor domains are exposed, flexible, and functionally engaged features that strongly predict where immunogenic extracellular epitopes are most likely located. </p>

```python
# Scrollable table for BioPlex Table
from pathlib import Path
import pandas as pd
from IPython.display import HTML, display

def _load_table_safely(path: str) -> pd.DataFrame:
    """
    Robust loader for CSV/TSV/Excel/plain tables.
    Tries delimiter inference and common encodings.
    """
    p = Path(path)
    if not p.exists():
        raise FileNotFoundError(f"File not found: {path}")
    
    # Excel
    if p.suffix.lower() in {".xlsx", ".xls"}:
        return pd.read_excel(p)
    
    # Try flexible CSV/TSV parsing
    try:
        return pd.read_csv(p, sep=None, engine="python")
    except UnicodeDecodeError:
        return pd.read_csv(p, sep=None, engine="python", encoding="latin-1")
    except Exception as e:
        # Fallbacks
        try:
            return pd.read_csv(p)  # assume comma
        except Exception:
            raise e

def show_scrollable_table(path: str, height: int = 480, width: str = "100%", sticky_header: bool = True, index: bool = False):
    """
    Render a pandas DataFrame inside a scrollable HTML container with an optional sticky header.
    """
    df = _load_table_safely(path)
    
    table_id = "scroll-table"
    html_table = df.to_html(index=index, escape=False, border=0, table_id=table_id)
    
    sticky_css = f"""
    #{table_id} thead th {{
        position: sticky;
        top: 0;
        background: #fafafa;
        z-index: 2;
        box-shadow: 0 1px 0 rgba(0,0,0,0.05);
    }}
    """ if sticky_header else ""
    
    style = f"""
    <style>
    .scroll-wrap {{
        max-height: {height}px;
        overflow: auto;           /* enables both vertical & horizontal scrolling */
        border: 1px solid #e5e7eb;
        border-radius: 10px;
        background: white;
    }}
    #{table_id} {{
        border-collapse: collapse;
        width: 100%;
        font-family: ui-sans-serif, system-ui, -apple-system, Segoe UI, Roboto, Helvetica, Arial, 'Apple Color Emoji','Segoe UI Emoji';
        font-size: 14px;
        table-layout: auto;
        white-space: nowrap;      /* prevent text wrap so horizontal scroll works */
    }}
    #{table_id} th, #{table_id} td {{
        padding: 8px 10px;
        border-bottom: 1px solid #f2f2f2;
        text-align: left;
    }}
    #{table_id} tbody tr:nth-child(even) {{ background: #fcfcfc; }}
    {sticky_css}
    </style>
    """
    
    container = f'<div class="scroll-wrap" style="width:{width}">{html_table}</div>'
    display(HTML(style + container))

# --- Use it on your uploaded file ---
show_scrollable_table(
    "GRIN1_Node-Table.csv",
    height=480,   # adjust height (px)
    width="100%", # or "900px"
    sticky_header=True,
    index=False
)
```

```python
# Automate Annotations

import pandas as pd
import requests
import time

# 1. Query UniProt REST API

def get_uniprot_id(gene_name, organism_id="9606"):
    """
    Return the reviewed (Swiss-Prot) UniProtKB accession for a gene,
    or None if nothing is found.
    """
    base_url = "https://rest.uniprot.org/uniprotkb/search"
    params = {
        "query": f"gene:{gene_name} AND organism_id:{organism_id} AND reviewed:true",
        "format": "json",
        "size": 1,
        "fields": "accession"
    }

    r = requests.get(base_url, params=params)
    if r.status_code != 200:
        print(f"[ID WARN] {gene_name}: HTTP {r.status_code}")
        return None

    data = r.json()
    results = data.get("results", [])
    if not results:
        print(f"[ID WARN] {gene_name}: no reviewed results")
        return None

    entry = results[0]
    acc = entry.get("primaryAccession")
    if not acc:
        print(f"[ID WARN] {gene_name}: no primaryAccession in result")
        return None

    return acc

# 3. Get Annotations using Go Query

def get_annotations_for_id(accession):
    """
    Given a UniProtKB accession, fetch TSV with:
    - Subcellular location [CC]
    - GO biological process
    - GO cellular component
    and return them as strings.
    """
    url = "https://rest.uniprot.org/uniprotkb/stream"
    params = {
        "query": f"accession:{accession}",
        "format": "tsv",
        "fields": "accession,cc_subcellular_location,go_p,go_c",
    }

    r = requests.get(url, params=params)
    if r.status_code != 200:
        print(f"[ANN WARN] {accession}: HTTP {r.status_code}")
        return {
            "Subcellular_Location": "",
            "GO_Biological_Process": "",
            "GO_Cellular_Component": "",
        }

    text = r.text.strip()
    lines = text.splitlines()
    if len(lines) < 2:
        print(f"[ANN WARN] {accession}: no annotation lines")
        return {
            "Subcellular_Location": "",
            "GO_Biological_Process": "",
            "GO_Cellular_Component": "",
        }

    header = lines[0].split("\t")
    row = lines[1].split("\t")
    data = dict(zip(header, row))

    sub = data.get("Subcellular location [CC]", "")
    go_bp = data.get("Gene Ontology (biological process)", "")
    go_cc = data.get("Gene Ontology (cellular component)", "")

    return {
        "Subcellular_Location": sub,
        "GO_Biological_Process": go_bp,
        "GO_Cellular_Component": go_cc,
    }

# Load node table
df = pd.read_csv("GRIN1_Node-Table.csv")

interactors = []

for _, row in df.iterrows():
    geneA = str(row["A.2"])
    geneB = str(row["B.2"])
    
    # If A is GRIN1 → partner = B
    if geneA.upper() == "GRIN1" and geneB != "":
        interactors.append(geneB)
    
    # If B is GRIN1 → partner = A
    if geneB.upper() == "GRIN1" and geneA != "":
        interactors.append(geneA)

# unique and explicitly drop GRIN1 if it snuck in
#interactors = sorted(set(g for g in interactors if g.upper() != "GRIN1"))
interactors = sorted(set(interactors))

print(interactors)

results = []

for g in genes:
    print(f"Processing {g}...")
    uid = get_uniprot_id(g)
    time.sleep(0.3)  # polite pause
    
    if uid is None:
        results.append({
            "Gene": g,
            "UniProt_ID": "",
            "Subcellular_Location": "",
            "GO_Biological_Process": "",
            "GO_Cellular_Component": "",
        })
        continue

    ann = get_annotations_for_id(uid)
    time.sleep(0.3)

    row = {
        "Gene": g,
        "UniProt_ID": uid,
        **ann,
    }
    results.append(row)

#annot_df = pd.DataFrame(results)
annot_df.to_csv("interactors_annotated.csv", index=False)
#annot_df
```

```python
import re

# --- Utility functions to clean text ---

def clean_location(text):
    """
    Extract meaningful compartments from UniProt location text.
    """
    keywords = [
        "cytoplasm", "cytosol", "cell membrane", "plasma membrane", 
        "endoplasmic reticulum", "golgi", "mitochondrion",
        "postsynapse", "synapse", "cell junction", "extracellular"
    ]
    text_low = text.lower()
    hits = [k for k in keywords if k in text_low]
    return "; ".join(sorted(set(hits)))

def clean_go_terms(bp, cc):
    """
    Extract high-level GO terms as short functional labels.
    """
    bp_low = bp.lower()
    cc_low = cc.lower()
    
    terms = []

    # ER / degradation / QC
    if "erad" in bp_low or "protein catabolic" in bp_low or "endoplasmic reticulum" in cc_low:
        terms.append("ER quality control")
    
    # cytoskeleton & migration
    if "actin" in bp_low or "migration" in bp_low or "adhesion" in bp_low:
        terms.append("cytoskeleton dynamics")
        
    # synapse / neuronal
    if "synapse" in bp_low or "postsynapse" in cc_low:
        terms.append("synaptic signaling")
        
    # autophagy
    if "autophagy" in bp_low:
        terms.append("autophagy")
    
    # trafficking
    if "transport" in bp_low or "vesicle" in bp_low:
        terms.append("trafficking")

    return "; ".join(sorted(set(terms)))

def classify_category(location, go_terms):
    """
    Classify into the three categories.
    """
    loc = location.lower()
    go = go_terms.lower()

    # ER / folding / degradation
    if "endoplasmic reticulum" in loc or "er quality" in go or "autophagy" in go:
        return "ER / Folding / ERAD / Autophagy"
    
    # synapse / cytoskeleton / membrane
    if ("synapse" in loc or "postsynapse" in loc or
        "cytoskeleton" in go or "membrane" in loc):
        return "Synaptic / Cytoskeletal / Membrane Scaffold"
    
    # nuclear / chromatin
    if "nucleus" in loc or "chromatin" in go:
        return "Nuclear / Chromatin / Transcription"
    
    return "Other / Needs Review"

def epitope_relevance(category):
    """
    Add a short explanation of why this interactor matters for NR1 epitope exposure.
    """
    if category.startswith("ER"):
        return "Interacts with NR1 during folding/QC → identifies structurally fragile extracellular regions likely to present epitopes."
    if category.startswith("Synaptic"):
        return "Binds intracellular motifs or organizes surface clusters → helps exclude intracellular regions as epitopes and highlight extracellular clustering."
    if category.startswith("Nuclear"):
        return "Not directly structural. Indicates signaling downstream of NR1 but not epitope-defining."
    return "Unclear relevance; review manually."

# --- Build cleaned table ---

clean_rows = []

for idx, row in annot_df.iterrows():
    loc = clean_location(row["Subcellular_Location"])
    go = clean_go_terms(row["GO_Biological_Process"], row["GO_Cellular_Component"])
    cat = classify_category(loc, go)
    epi = epitope_relevance(cat)

    clean_rows.append({
        "Gene": row["Gene"],
        "UniProt_ID": row["UniProt_ID"],
        "Location": loc,
        "Function_Keywords": go,
        "Category": cat,
        "Epitope_Relevance": epi
    })

clean_df = pd.DataFrame(clean_rows)
clean_df.to_csv("interactors_readable.csv", index=False)
clean_df
```

   <h2 style="color:#2c3e50; font-family:'Helvetica Neue',sans-serif; font-size:28px; margin-bottom:20px;">
    Interaction-Based Interpretation of NR1 Epitope Exposure
  </h2>

  <p style="color:#34495e; font-family:'Arial',sans-serif; font-size:15px; line-height:1.6; text-indent:25px;">
    To understand which structural regions of NR1 are most likely to contain immunogenic epitopes, 
    I examined the subcellular locations and biological functions of each GRIN1 interactor in the BioPlex network.
    This allows us to infer whether each partner necessarily contacts <b>extracellular</b> domains of NR1, 
    where antibody epitopes reside, or whether it binds <b>intracellular</b> or <b>ER-luminal</b> regions, 
    which helps exclude those surfaces as epitope candidates.
  </p>

  <h3 style="color:#2c3e50; font-family:'Helvetica Neue',sans-serif; font-size:22px; margin-top:25px;">
    1. Partners Implicating Extracellular Epitope Surfaces
  </h3>

  <p style="color:#34495e; font-family:'Arial',sans-serif; font-size:15px; line-height:1.6; text-indent:25px;">
    Several interactors are associated with <b>endoplasmic reticulum quality control and ER-associated degradation (ERAD)</b>, 
    including <b>FBXO2</b>, <b>RETREG3</b>, and <b>ANKRD13C</b>. 
    These proteins typically recognize or process <i>misfolded or partially folded regions</i> of NR1 while it transits through the ER. 
    Because the ER lumen corresponds topologically to the <b>extracellular</b> side of the mature receptor, 
    these interactions imply that specific NR1 extracellular domains require extensive folding oversight. 
    Such conformationally sensitive, surface-exposed regions are precisely the kinds of structures known to form <b>conformational antibody epitopes</b>.
    Thus, the presence of ERQC/ERAD partners strongly supports the idea that NR1’s ATD and LBD surfaces contain 
    epitope-prone structural motifs.
  </p>

  <h3 style="color:#2c3e50; font-family:'Helvetica Neue',sans-serif; font-size:22px; margin-top:25px;">
    2. Partners Indicating Intracellular Binding (Not Accessible to Antibodies)
  </h3>

  <p style="color:#34495e; font-family:'Arial',sans-serif; font-size:15px; line-height:1.6; text-indent:25px;">
    In contrast, cytoskeletal and synaptic scaffold proteins such as <b>DOCK1</b>, <b>ELMO1</b>, and <b>ELMO2</b> 
    function primarily at the <b>cytoplasmic</b> face of membrane receptors. 
    These interactions are consistent with binding to <i>intracellular loops or the NR1 C-terminal domain</i>. 
    Importantly, these surfaces are <b>not accessible</b> to circulating IgG in autoimmune conditions. 
    Therefore, identifying these intracellular partners helps refine epitope mapping by ruling out 
    intracellular regions as potential antibody targets and focusing attention on extracellular structural domains.
  </p>

  <p style="color:#34495e; font-family:'Arial',sans-serif; font-size:15px; line-height:1.6; text-indent:25px;">
    Altogether, the pattern of interactors clearly stratifies NR1 structural regions into (1) extracellular, 
    conformationally vulnerable domains supported by ERQC/ERAD interactions and (2) intracellular scaffolding interfaces 
    supported by cytoskeletal regulators. This division strongly reinforces that <b>NR1 antibody epitopes most likely reside 
    on the extracellular ATD/LBD surfaces</b>, where folding stress and partner engagement converge.
  </p>

<b>RUN domain</b> is linked to small GTPase signaling and vesicular trafficking, which fits with GRIN1’s involvement in synaptic vesicle recycling and trafficking.
<br><br>
<b>AAA domain</b> is characteristic of ATPase-associated proteins involved in protein unfolding, degradation, and trafficking, suggesting possible coupling to chaperone or proteostasis pathways that regulate receptor assembly.<br><br>
</p>

<!-- ANALYSIS SECTION -->
<div style="background-color:#f8f9fb; padding:25px; border-radius:12px; box-shadow:0px 4px 8px rgba(0,0,0,0.08);">

  <h2 style="color:#2c3e50; font-family:'Helvetica Neue',sans-serif; font-size:28px; margin-bottom:20px;">
    Analysis
  </h2>

  <p style="color:#34495e; font-family:'Arial',sans-serif; font-size:15px; line-height:1.6;">
<b> Integrating NR1 Structural Architecture with BioPlex Domain–Domain Interactions</b><br>
NR1 has three major structural regions relevant for interaction mapping. The amino-terminal domain (ATD) and ligand-binding domain (LBD; S1 and S2) are extracellular. The ligand-gated ion channel (LGIC) domain forms the transmembrane pore, and the C-terminal domain (CTD) is intracellular. This topology allows interactors to be assigned to likely binding surfaces based on their domain composition and cellular localization.
<br><br>
BioPlex domain–domain enrichment in the NR1 interactome identifies three main domain types among NR1 partners: AAA ATPase domains, RUN domains, and bromodomains. These domains correspond to different cellular functions and therefore indicate different modes of NR1 engagement.
<br><br>
AAA ATPase and ERAD-associated proteins (e.g., FBXO2) interact with NR1 during folding in the ER lumen. Because the ER lumen is topologically equivalent to the extracellular space, these interactions necessarily involve extracellular domains such as ATD, LBD, and nearby extracellular loops. Regions requiring frequent folding surveillance are typically flexible or conformationally dynamic, which makes them common sites for antibody epitopes. These data support the presence of extracellular conformational epitopes on NR1, consistent with known properties of anti-NMDAR antibodies.
<br><br>
RUN-domain proteins are involved in small GTPase signaling, internalization, and trafficking. These proteins bind the cytosolic side of membrane receptors, usually interacting with short intracellular motifs. RUN-domain partners therefore map to the NR1 CTD or intracellular loops, which are not exposed to circulating antibodies. Their presence helps exclude intracellular regions from consideration as epitope sites.
<br><br>
Bromodomain-containing proteins participate in chromatin remodeling and transcriptional regulation. They do not bind NR1 directly but indicate downstream signaling pathways linked to NR1 activity. Although they do not identify structural binding sites, their association suggests NR1 participates in stable signaling assemblies, which can increase receptor clustering and surface density. This indirectly contributes to epitope accessibility but does not define epitope location.
<br><br>
Overall, combining NR1’s structural map with BioPlex domain enrichments yields a consistent interpretation:<br>
• AAA/ERAD partners point to extracellular folding-sensitive domains as likely epitope regions.<br>
• RUN-domain partners indicate intracellular regions that cannot serve as epitopes.<br>
• Bromodomain partners reflect signaling context but do not locate epitopes.<br>
<br><br>
These findings support the conclusion that NR1 epitopes are located on the extracellular ATD/LBD.
  </p>
